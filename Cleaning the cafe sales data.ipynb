```python
import pandas as pd ##Importing Pandas
import numpy as np ##Importing Numpy
import warnings ##importing warnings
warnings.filterwarnings("ignore") ##Initiating to ignore the warnings
```

```python
df = pd.read_csv("dirty_cafe_sales.csv") ##Reading the csv into a DF
```

```python
print(df.shape) ##Getting the shape
df.head() ##Looking at the data
```

```python
df.isna().sum().sort_values() ##Checking the null values in data
```

```python
df.replace({'ERROR':np.nan,'UNKNOWN':np.nan}) ##Replacing the wrong values to nan(imported NUMPY for this purpose)
```

```python
df.replace({'ERROR':np.nan,'UNKNOWN':np.nan},inplace=True) ##Replacing the wrong values to nan(imported NUMPY for this purpose), missed to mention inplace in above step
```

```python
df.isna().sum().sort_values() ##The actual count of nan's in the data
```

```python
twofield = df[['Item','Price Per Unit']] ##Creating a df with Item and Price
```

```python
twofield.shape ##Checking the shape
```

```python
twofield.drop_duplicates(inplace=True) ##Dropping the duplicates
```

```python
twofield.dropna(inplace=True) ##Dropping the null records
```

```python
twofield.shape ##checking the shape after the drop of nulls and duplicates
```

```python
df1 = pd.merge(df,twofield,on='Item',how='left') ##Joining the twofield dataframe to the df and creating a new df: df1
```

```python
df1.shape ##checking the shape
```

```python
df1.head() ##to look how the df looks
```

```python
df1.isna().sum().sort_values()  ##To see the nulls after adding the new Price column
```

```python
df1['Price Per Unit_y'] = df1['Price Per Unit_y'].fillna(df1['Price Per Unit_x']) ##Filling the nulls in new column
```

```python
df1.isna().sum().sort_values() ##checking the nulls after fillna
```

```python
df1.drop(['Price Per Unit_x'],axis='columns',inplace=True) ##Dropping the actual column
```

```python
df1.rename(columns={'Price Per Unit_y':'Price Per Unit'},inplace=True) ##Renaming the new column
```

```python
df1.shape ##checking the shape
```

```python
df1.head() ##Seeing the df
```

```python
threefield = df1[df1['Price Per Unit'].isin([np.nan])][['Transaction ID','Quantity','Total Spent']] ##Creating a df with T ID, Quant, T Spent records where the Price column is null
```

```python
print(threefield.shape) ##Checking the shape and seeing the df
threefield.head()
```

```python
threefield.dropna(inplace=True) ##Dropping the null values from the df
```

```python
threefield.shape ##Checking the shape after dropna
```

```python
threefield['Quantity'] = threefield['Quantity'].astype(float) ##Changing the dtypes for the Quantity, Total Spent field
threefield['Total Spent'] = threefield['Total Spent'].astype(float)
```

```python
threefield['PPU'] = threefield['Total Spent']/threefield['Quantity'] ##creating a new price column by calculating the Quantity and Total Spent
```

```python
print(threefield.shape) ##Checking the shape and looking at the data
threefield.head()
```

```python
threefield.drop(['Quantity','Total Spent'],axis='columns',inplace=True) ##Dropping the field that are not required
```

```python
threefield.head() ##looking at the df sturc
```

```python
df1 = pd.merge(df1,threefield,on='Transaction ID',how='left') ##Joining the price field to the main df1 based on T ID
```

```python
df1.isna().sum().sort_values() ##Now checking the nulls
```

```python
df1['PPU'] = df1['PPU'].fillna(df1['Price Per Unit']) ##Filling the nulls in new column with values from Price Per Unit
```

```python
df1.isna().sum().sort_values() ##checking the nulls after the fillna
```

```python
df1.drop(['Price Per Unit'],axis='columns',inplace=True) ##Dropping the actual field
```

```python
df1.rename(columns={'PPU':'Price Per Unit'},inplace=True) ##renaming the new filed with less nulls
```

```python
print(df1.shape)
df1.isna().sum().sort_values() ##checking the shape and looking at the count of nulls in df
```

```python
df1.dtypes ##checking the dtypes of the df1
```

```python
df1['Price Per Unit'] = df1['Price Per Unit'].astype(float)
df1['Quantity'] = df1['Quantity'].astype(float)
df1['Total Spent'] = df1['Total Spent'].astype(float) ##Converting all the numerical field dtype to be float
```

```python
df1.dtypes ##checking dtypes after the change
```

```python
df1['QNew'] = df1['Quantity'].fillna(df1['Total Spent']/df1['Price Per Unit'])
df1['TSNew'] = df1['Total Spent'].fillna(df1['QNew']*df1['Price Per Unit']) ##Creating new Quantity and Total Spent fields based on logics
```

```python
print(df1.shape)
df1.head() ##Checking the shape and looking the struc of df1
```

```python
df1.isna().sum().sort_values() ##Now looking at the nulls after the new Quanitty and Total Spet fields
```

```python
df1.drop(['Total Spent','Quantity'],axis='columns',inplace=True) ##Dropping the original field which has more nulls
```

```python
df1.head()
```

```python
df1.rename(columns={'QNew':'Quantity'.{'TSNew':'Total Spent'}},inplace=True) ##Renaming the new columns to actual names
```

```python
df1.head() ##Looking at the df1 after the change
```

```python
df1.median() ##In the numerical field we're still having nulls, so finding the median to fill them
```

```python
df1['Price Per Unit'] = df1['Price Per Unit'].fillna(df1['Price Per Unit'].median()) ##Finding the median of Price and filling in na's
df1['Quantity'] = df1['Quantity'].fillna(df1['Quantity'].median()) ##Finding the median of Quantity and filling the na's
df1['Total Spent'] = df1['Total Spent'].fillna(9) ##Filling the na's in Total Spent with 9
```

```python
df1.isna().sum().sort_values() ##checking the count of na's after the above three steps
```

```python
GroupByPpu = df1.groupby(df1['Price Per Unit']) ##creating a group by object
```

```python
df1['ItemNew'] = GroupByPpu.Item.transform(lambda x: x.fillna(x.mode()[0])) ##Filling the na's in Item field with mode of each Price
```

```python
print(df1.shape)
print(df1.head())
df1.isna().sum().sort_values() ##Checking the shape, looking at the data and seeing the na's
```

```python
df1.drop(['Item'],axis='columns',inplace=True) ##Dropping the actual Item field
```

```python
df1.rename(columns={'ItemNew':'Item'},inplace=True) ##Renaming the new field
```

```python
print(df1.shape)
```

```python
df1 = df1.sort_values(by='Transaction ID') ##Sorting the data by T ID
```

```python
df1.head() ##Checking the df after the sort
```

```python
df1['TDNew'] = df1['Transaction Date'].fillna(method='ffill') ##Filling the nulls in Date as per the previous date ordered by T ID
```

```python
df1.head() ##checking the data struc to see the change happened
```

```python
df1.drop(['Transaction Date'],axis='columns',inplace=True) ##Dropping the actual date field
df1.rename(columns={'TDNew':'Transaction Date'},inplace=True) ##Renaming the new date field to actual name
```

```python
df1.head() ##checking the data after the change
```

```python
df1.isna().sum().sort_values() #checking for the nulls values
```

```python
GroupByItem = df1.groupby(df1['Item']) ##creating a groupby object based on the Item
```

```python
df1['PMNew'] = GroupByItem['Payment Method'].transform(lambda x: x.fillna(x.mode()[0])) ##Fillng the na's in Pay Method based on the mode of each Item
df1['LNew'] = GroupByItem['Location'].transform(lambda x: x.fillna(x.mode()[0]))##Fillng the na's in Location based on the mode of each Item
```

```python
df1.head() ##Looking at the data after the change
```

```python
df1.isna().sum() ##checking the nuls after the fillna
```

```python
df1.drop(['Payment Method','Location'],axis='columns',inplace=True) ##Dropping the actual columns which have nulls
```

```python
df1.rename(columns={'PMNew':'Payment Method','LNew':'Location'},inplace=True) ##Renaming the new fields
```

```python
df1.head() ##checking the data
```

```python
df1.isna().sum() ##Checking the nulls in df
```

```python
df1 = df1.reset_index(drop=True) ##Resertting the index as we have changed the sort earlier
```

```python
df1 ##checking the df
```

```python
df1.to_csv('CleanedCafeSalesData.csv') ##Saving the df to a csv
```
